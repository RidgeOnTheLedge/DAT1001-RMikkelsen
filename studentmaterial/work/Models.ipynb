## The case of Robert Swain

```python
population_proportions = make_array(.26, .74)
population_proportions 
```

```python
def panel_proportion():
    return sample_proportions(100, population_proportions).item(0)
```

```python
proportion_counts = []

for i in np.arange(10000):
    proportion_counts.append(panel_proportion() * 100)
```

```python
samples_table = Table().with_columns("Number of Black Men of 100", proportion_counts)

```

```python
proportion_counts
```

```python
samples_table.hist(bins=np.arange(4.5, 45))

plt.ylim(-0.002, 0.09)
```

## Mendel's pea planet model

```python
observerd_purples = 709 / 929
```

```python
def purple_flowers():
    p = make_array(.75, .25)
    return abs(sample_proportions(929, p).item(0) - 0.75)
```

```python

purple_diff = []

for i in np.arange(10000):
    purple_diff.append(purple_flowers())
```

```python
purple_tb = Table().with_columns("Absolute difference if the model is true", purple_diff)
```
