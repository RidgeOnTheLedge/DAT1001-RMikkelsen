```python
def name_and_age(name, year):
    '''
    Returns a string with name and age of person.
    Arguments:
        name: the name of the person
        year: the year the person was born
    Returns:
        string with name and age.
    '''
    return name + ' is ' + str(age(year)) + ' years old.' 
```

```python
name_and_age('Ridge', 2006)
```
