```python
def sim_flips():
    hundred_tosses = np.random.choice(coin, 100)
    num_heads = np.count_nonzero(hundred_tosses == 'Heads')
    return num_heads
```

```python
flip_list = []
rep_list = []
for i in np.arange(10000):
    
    num = sim_flips()
    flip_list.append(num)
    rep_list.append(i)
```

```python
sim_results = Table().with_columns('Repetition', rep_list,
                                  'Num of Heads', flip_list)
sim_results.column('Num of Heads')
sim_results
```

```python
import matplotlib.pyplot as plt

x_values = sim_results.column('Num of Heads')
y_values = sim_results.column('Repetition')

plt.bar(x_values, y_values)
# plt.show()  <-- Remove or comment this out
plt.savefig('my_plot.png')  # Saves the image to your project folder
```
